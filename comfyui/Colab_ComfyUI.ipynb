```python
COMFYUI_PATH = '/content/ComfyUI'

LINKS = {
    'models': f'{GDRIVE_BASE}/models',
    'custom_nodes': f'{GDRIVE_BASE}/custom_nodes',
    'input': f'{GDRIVE_BASE}/input',
    'output': f'{GDRIVE_BASE}/output',
    'temp': f'{GDRIVE_BASE}/temp',
    'user': f'{GDRIVE_BASE}/user'
}

for name, target in LINKS.items():
    source = os.path.join(COMFYUI_PATH, name)
    # Remove old local folder
    if os.path.islink(source) or os.path.isdir(source):
        !rm -rf "{source}"
    # Create symlink
    os.symlink(target, source)

print("✅ ComfyUI is now fully configured to use Google Drive.")
```

```python
import os
import subprocess

# COMFYUI MODEL CONFIGURATION

COMFYUI_MODELS = "/content/drive/MyDrive/ComfyUI/models"

CHECKPOINT_DIR = os.path.join(COMFYUI_MODELS, "checkpoints")
UPSCALE_DIR = os.path.join(COMFYUI_MODELS, "upscale_models")

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(UPSCALE_DIR, exist_ok=True)

print("✅ Model directories ready.")
print(f"Checkpoint directory: {CHECKPOINT_DIR}")
print(f"Upscaler directory:   {UPSCALE_DIR}")


# 1. SDXL BASE CHECKPOINT

SDXL_FILENAME = "sd_xl_base_1.0.safetensors"

SDXL_URL = (
    "https://huggingface.co/stabilityai/"
    "stable-diffusion-xl-base-1.0/resolve/main/"
    "sd_xl_base_1.0.safetensors"
)

SDXL_PATH = os.path.join(CHECKPOINT_DIR, SDXL_FILENAME)


if not os.path.exists(SDXL_PATH):

    print("\n⬇️ Downloading SDXL Base...")

    subprocess.run([
        "wget",
        "-O",
        SDXL_PATH,
        SDXL_URL
    ], check=True)

    print("✅ SDXL Base downloaded.")

else:

    print("✅ SDXL Base already exists.")


# REAL-ESRGAN UPSCALER

UPSCALE_FILENAME = "RealESRGAN_x4plus.pth"

# Correct official Real-ESRGAN model URL
UPSCALE_URL = (
    "https://github.com/xinntao/Real-ESRGAN/"
    "releases/download/v0.1.0/"
    "RealESRGAN_x4plus.pth"
)

UPSCALE_PATH = os.path.join(
    UPSCALE_DIR,
    UPSCALE_FILENAME
)

if not os.path.exists(UPSCALE_PATH):

    print("\n⬇️ Downloading RealESRGAN x4 upscaler...")

    subprocess.run([
        "wget",
        "-O",
        UPSCALE_PATH,
        UPSCALE_URL
    ], check=True)

    print("✅ RealESRGAN upscaler downloaded.")

else:

    print("✅ RealESRGAN upscaler already exists.")


# VERIFY MODELS - I included this section so that I could check whether the models are loaded properly

print("\n" + "=" * 60)
print("MODEL VERIFICATION")
print("=" * 60)

print(
    f"SDXL Base: "
    f"{'✅ FOUND' if os.path.exists(SDXL_PATH) else '❌ MISSING'}"
)

print(
    f"RealESRGAN: "
    f"{'✅ FOUND' if os.path.exists(UPSCALE_PATH) else '❌ MISSING'}"
)

print("\n📁 Checkpoint files:")
for file in os.listdir(CHECKPOINT_DIR):
    print("  -", file)

print("\n📁 Upscaler files:")
for file in os.listdir(UPSCALE_DIR):
    print("  -", file)
```

```python

# FINAL COMFYUI MODEL CHECK - This was included to double check the models cause I ran into a situation where the model was not downloaded

required_models = {
    "SDXL Base":
        "/content/drive/MyDrive/ComfyUI/models/checkpoints/sd_xl_base_1.0.safetensors",

    "RealESRGAN x4":
        "/content/drive/MyDrive/ComfyUI/models/upscale_models/RealESRGAN_x4plus.pth",
}

print("\n🔍 Checking required models...\n")

all_models_available = True

for name, path in required_models.items():

    if os.path.isfile(path):

        size_mb = os.path.getsize(path) / (1024 * 1024)

        print(f"✅ {name}")
        print(f"   {size_mb:.1f} MB")
        print(f"   {path}\n")

    else:

        print(f"❌ {name}")
        print(f"   Missing: {path}\n")

        all_models_available = False


if all_models_available:
    print("🎉 All required models are available.")
else:
    print("⚠️ One or more required models are missing.")
```

```python
import subprocess
import time
import requests

# Start ngrok tunnel to port 8188
ngrok_process = subprocess.Popen(['./ngrok', 'http', '8188'],
                                 stdout=subprocess.DEVNULL,
                                 stderr=subprocess.STDOUT)

# Wait and retry connection
for i in range(10):
    try:
        r = requests.get('http://localhost:4040/api/tunnels')
        public_url = r.json()['tunnels'][0]['public_url']
        print(f"✅ Ngrok tunnel established: {public_url}")
        break
    except Exception as e:
        print(f"⏳ Attempt {i+1}/10: Ngrok not ready yet...")
        time.sleep(2)
else:
    print("❌ Ngrok failed to start. Please restart the runtime and try again.")

```
